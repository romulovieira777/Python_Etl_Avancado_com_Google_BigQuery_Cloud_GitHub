```python
from google.cloud import bigquery
import pandas as pd
```

```python
# Especifique o caminho para o arquivo JSON da chave de serviço
credencial = r'caminho\arquivo.json'

# Crie um cliente do BigQuery usando as credenciais do arquivo JSON
client = bigquery.Client.from_service_account_json(credencial)
```

```python
# Construa sua consulta SQL
query = """
SELECT
    id
  , first_name
  , last_name
  , email
FROM
    `pythoncurso-398511.AulaPython.Clientes`
WHERE
    state IN ('Piauí', 'Alagoas', 'Amazonas', 'Maranhão', 'Rondonia')
"""

# Execute a consulta
resultado = client.query(query)

# Converta o resultado para um DataFrame do Pandas
df = resultado.to_dataframe()
df.head()
```
